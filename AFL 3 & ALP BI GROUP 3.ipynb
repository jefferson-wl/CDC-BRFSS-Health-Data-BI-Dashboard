```python
import pandas as pd

# Load the dataset
file_path = r"c:\Users\ASUS\Downloads\Nutrition,_Physical_Activity,_and_Obesity_-_Behavioral_Risk_Factor_Surveillance_System_202512.csv"
df = pd.read_csv(file_path)

# Check the first few rows to make sure it loaded correctly
print(df.head())
```

```python
# Display the first 5 rows
print("First 5 rows")
display(df.head())
```

```python
# Check data types and column names
print("Data Info")
df.info()
```

```python
# Check for null values in each column
print("Null Value Count")
print(df.isnull().sum())
```

```python
# List of columns to drop
cols_to_drop = [
    'Data_Value_Alt',              # Duplicate of Data_Value
    'Data_Value_Unit',             # Mostly null / implied
    'Data_Value_Footnote_Symbol',  # Metadata for null values
    'Data_Value_Footnote',         # Metadata for null values
    'Total', 'Age(years)',         # Sparse columns (info is in Stratification columns)
    'Education', 'Sex', 
    'Income', 'Race/Ethnicity' 
]

# Create a clean copy by dropping these columns
df_clean = df.drop(columns=cols_to_drop)

print(f"Dropped columns: {cols_to_drop}")
```

```python
# Count rows before dropping
rows_before = len(df_clean)
```

```python
# Drop rows where 'Data_Value' is NaN
df_clean = df_clean.dropna(subset=['Data_Value'])
```

```python
# Calculate dropped rows
rows_after = len(df_clean)
print(f"Rows before: {rows_before}")
print(f"Rows after: {rows_after}")
print(f"Dropped {rows_before - rows_after} rows with missing data values.")
```

```python
# Remove commas and convert Sample_Size to float
# We use .astype(str) first to handle any potential non-string types safely
df_clean['Sample_Size'] = df_clean['Sample_Size'].astype(str).str.replace(',', '').astype(float)

# Verify the change
print("Sample_Size data type is now:", df_clean['Sample_Size'].dtype)
display(df_clean[['LocationDesc', 'Data_Value', 'Sample_Size']].head())
```

```python
# Reset index to have continuous row numbers
df_clean = df_clean.reset_index(drop=True)

# Final check of the clean dataframe
print("Final Cleaned Data Info")
df_clean.info()

# df_clean.to_csv('Cleaned_Nutrition_Data.csv', index=False)
# print("Cleaned dataset saved to 'Cleaned_Nutrition_Data.csv'")
```

```python
# Check for null values in the cleaned dataset
print("Null Values After Cleaning")
null_counts = df_clean.isnull().sum()
print(null_counts)

# Optional: Visualize if there are any remaining nulls
if null_counts.sum() == 0:
    print("\nSuccess: No missing values remaining in the dataset!")
else:
    print("\nNote: There are still some missing values (e.g., in GeoLocation).")
    # Show columns that still have nulls
    print(null_counts[null_counts > 0])
```

```python
# Filter out rows with missing GeoLocation (removes 'National' level data)
df_clean = df_clean.dropna(subset=['GeoLocation'])

# Reset the index again after dropping rows
df_clean = df_clean.reset_index(drop=True)

# Verify that there are no nulls left in GeoLocation
print(f"Nulls in GeoLocation: {df_clean['GeoLocation'].isnull().sum()}")
print(f"Unique locations remaining: {df_clean['LocationAbbr'].nunique()}")
```

```python
# Save the final cleaned dataset
df_clean.to_csv('Cleaned_Nutrition_Data_Map_Ready.csv', index=False)
print("File saved successfully as 'Cleaned_Nutrition_Data_Map_Ready.csv'")
```
